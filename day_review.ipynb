```python
import os
from datetime import date
```

```python
def files_func(folder_path = 'transcripts/'):
    # Ensure the folder exists
    if not os.path.exists(folder_path):
        print(f"The folder {folder_path} does not exist.")
        return

    file_list = []
    files_names = []
    # Iterate through the generator returned by os.walk()
    for root, dirs, files in os.walk(folder_path):
        file_list.append(root)
        files_names.append(files)

    return file_list, file_names
```

```python
def new_day(date, new_file):
    with open(new_file, "a") as file:  
        file.write("Day Review" + "\n")
        file.write("date : " + date + "\n")
```

```python
def read_file_line_by_line(new_file, old_file, name):
    with open(new_file, "a") as file:  
        file.write(name + "\n")
    
        with open(old_file, 'r') as file_old:  
            for line in file_old:  
                file.write(* + line + "\n")
                  
```
